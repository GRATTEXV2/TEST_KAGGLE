```python
import pandas as pd
df = pd.read_csv(r"C:\Users\gratterif\Desktop\Python\TestGithub\TEST_KAGGLE\job_salary_prediction_dataset.csv")

df
```

```python
#Create a barchart to show the distribution of job titles in the dataset
import matplotlib.pyplot as plt     
job_title_counts = df['job_title'].value_counts()
plt.figure(figsize=(10,6))
job_title_counts.plot(kind='bar')
plt.title('Distribution of Job Titles')
plt.xlabel('Job Title')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()  
```

```python
#Mostrami per ogni industries nella stessa griglia più grafici che mostrano l'andamento delle metriche numeriche (ad esempio, salary) in base a un'altra variabile (ad esempio, experience_level)
import seaborn as sns
plt.figure(figsize=(12, 8))
sns.boxplot(x='experience_years', y='salary', hue='industry', data=df)
plt.title('Salary Distribution by Experience Level and Industry')
plt.xlabel('Experience Level')
plt.ylabel('Salary')
plt.legend(title='Industry')
plt.show()
```
