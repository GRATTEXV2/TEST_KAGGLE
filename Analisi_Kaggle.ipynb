```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 110


print("=== Dataset Shape ===")
print(f"Rows: {df.shape[0]:,}  |  Columns: {df.shape[1]}\n")

print("=== Column Data Types ===")
print(df.dtypes.to_string())

print("\n=== Missing Values ===")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.any() else "No missing values found.")

print("\n=== Descriptive Statistics (Numerical) ===")
df.describe().T
```

```python
# Numerical Features Distribution Grid

import math

num_cols = df.select_dtypes(include="number").columns.tolist()

n_cols = 3
n_rows = math.ceil(len(num_cols) / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(6 * n_cols, 4 * n_rows))
fig.suptitle("Numerical Features — Distribution Grid", fontsize=16, fontweight="bold", y=1.01)

axes = axes.flatten()
colors = sns.color_palette("muted", len(num_cols))

for i, (col, color) in enumerate(zip(num_cols, colors)):
    ax = axes[i]
    # Histogram + KDE overlay
    ax.hist(df[col], bins=40, color=color, edgecolor="white", alpha=0.6, label="Histogram")
    sns.kdeplot(df[col], ax=ax, color=color, linewidth=2, label="KDE")
    # Vertical lines for mean and median
    ax.axvline(df[col].mean(),   color="red",    linestyle="--", linewidth=1.2, label=f"Mean: {df[col].mean():.1f}")
    ax.axvline(df[col].median(), color="orange", linestyle=":",  linewidth=1.2, label=f"Median: {df[col].median():.1f}")
    ax.set_title(col.replace("_", " ").title(), fontsize=12, fontweight="bold")
    ax.set_xlabel(col)
    ax.set_ylabel("# Observations")
    ax.legend(fontsize=7)
    # Grid styling
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.set_facecolor("#f9f9f9")

# Hide unused subplots
for j in range(len(num_cols), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

## 1. Target Variable — Salary Distribution

```python

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Salary Distribution", fontsize=15, fontweight="bold")

# Histogram + KDE
axes[0].hist(df["salary"], bins=60, color="#4C72B0", edgecolor="white", alpha=0.85)
axes[0].set_title("Histogram")
axes[0].set_xlabel("Salary")
axes[0].set_ylabel("Count")

# KDE
sns.kdeplot(df["salary"], ax=axes[1], fill=True, color="#4C72B0", linewidth=2)
axes[1].set_title("KDE Plot")
axes[1].set_xlabel("Salary")

# Box plot
axes[2].boxplot(df["salary"], vert=False, patch_artist=True,
                boxprops=dict(facecolor="#4C72B0", color="#4C72B0", alpha=0.7),
                medianprops=dict(color="white", linewidth=2))
axes[2].set_title("Box Plot")
axes[2].set_xlabel("Salary")
axes[2].set_yticks([])

plt.tight_layout()
plt.show()

print(f"Mean Salary: ${df['salary'].mean():,.0f}")
print(f"Median Salary: ${df['salary'].median():,.0f}")
print(f"Std Dev: ${df['salary'].std():,.0f}")
print(f"Min: ${df['salary'].min():,.0f}  |  Max: ${df['salary'].max():,.0f}")
print(f"Skewness: {df['salary'].skew():.3f}  |  Kurtosis: {df['salary'].kurt():.3f}")
```

## 2. Numerical Features — Distributions & Correlation

```python

#num_cols = ["experience_years", "skills_count", "certifications", "salary"]

num_cols = df.select_dtypes(include="number").columns.tolist()

print("le colonne numeriche sono:", ", ".join(num_cols))

fig, axes = plt.subplots(2, len(num_cols), figsize=(18, 8))
fig.suptitle("Numerical Features — Distributions", fontsize=14, fontweight="bold")

colors = sns.color_palette("muted", len(num_cols))

for i, (col, color) in enumerate(zip(num_cols, colors)):
    # Histogram
    axes[0, i].hist(df[col], bins=40, color=color, edgecolor="white", alpha=0.85)
    axes[0, i].set_title(col.replace("_", " ").title())
    axes[0, i].set_xlabel(col)
    axes[0, i].set_ylabel("Count")
    # Box plot
    axes[1, i].boxplot(df[col], vert=True, patch_artist=True,
                       boxprops=dict(facecolor=color, alpha=0.7),
                       medianprops=dict(color="white", linewidth=2))
    axes[1, i].set_title(f"{col.replace('_', ' ').title()} — Box")
    axes[1, i].set_xticks([])

plt.tight_layout()
plt.show()
#
```

```python

# Correlation heatmap
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(7, 5))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="coolwarm",
            vmin=-1, vmax=1, linewidths=0.5, ax=ax, square=True,
            cbar_kws={"shrink": 0.8})
ax.set_title("Correlation Matrix — Numerical Features", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 3. Categorical Features — Distributions

```python

cat_cols = ["job_title", "education_level", "industry", "company_size", "remote_work"]

#cat_cols = df.select_dtypes(include="string").columns.tolist()
fig, axes = plt.subplots(2, 3, figsize=(20, 11))
fig.suptitle("Categorical Features — Value Counts", fontsize=14, fontweight="bold")
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    counts = df[col].value_counts()
    bars = axes[i].bar(counts.index, counts.values,
                       color=sns.color_palette("muted", len(counts)))
    axes[i].set_title(col.replace("_", " ").title(), fontsize=12)
    axes[i].set_xlabel("")
    axes[i].set_ylabel("Count")
    axes[i].tick_params(axis="x", rotation=35)
    for bar in bars:
        axes[i].text(bar.get_x() + bar.get_width() / 2,
                     bar.get_height() + counts.max() * 0.01,
                     f"{int(bar.get_height()):,}",
                     ha="center", va="bottom", fontsize=7)

# Use last subplot for location top-10
loc_counts = df["location"].value_counts().head(10)
bars = axes[5].bar(loc_counts.index, loc_counts.values,
                   color=sns.color_palette("muted", 10))
axes[5].set_title("Top 10 Locations", fontsize=12)
axes[5].set_ylabel("Count")
axes[5].tick_params(axis="x", rotation=35)
for bar in bars:
    axes[5].text(bar.get_x() + bar.get_width() / 2,
                 bar.get_height() + loc_counts.max() * 0.01,
                 f"{int(bar.get_height()):,}",
                 ha="center", va="bottom", fontsize=7)

plt.tight_layout()
plt.show()
```

## 4. Salary vs Categorical Features

```python

# Salary by Job Title — horizontal box plots ordered by median
order = (df.groupby("job_title")["salary"]
           .median()
           .sort_values(ascending=False)
           .index)

fig, ax = plt.subplots(figsize=(14, 7))
sns.boxplot(data=df, y="job_title", x="salary", order=order,
            palette="muted", ax=ax, linewidth=0.8)
ax.set_title("Salary Distribution by Job Title", fontsize=13, fontweight="bold")
ax.set_xlabel("Salary")
ax.set_ylabel("")
plt.tight_layout()
plt.show()
```

```python

# Salary vs Education, Industry, Company Size, Remote Work — violin plots
cat_salary = ["education_level", "industry", "company_size", "remote_work"]
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.suptitle("Salary Distribution by Categorical Features", fontsize=14, fontweight="bold")
axes = axes.flatten()

for i, col in enumerate(cat_salary):
    order_col = (df.groupby(col)["salary"]
                   .median()
                   .sort_values(ascending=False)
                   .index)
    sns.violinplot(data=df, x=col, y="salary", order=order_col,
                   palette="muted", ax=axes[i], inner="quartile", linewidth=0.8)
    axes[i].set_title(f"Salary by {col.replace('_', ' ').title()}", fontsize=12)
    axes[i].set_xlabel("")
    axes[i].set_ylabel("Salary")
    axes[i].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

```python

# Mean salary by location — top 15
top_locations = (df.groupby("location")["salary"]
                   .mean()
                   .sort_values(ascending=False)
                   .head(15))

fig, ax = plt.subplots(figsize=(14, 5))
bars = ax.bar(top_locations.index, top_locations.values,
              color=sns.color_palette("muted", 15))
ax.set_title("Average Salary — Top 15 Locations", fontsize=13, fontweight="bold")
ax.set_xlabel("Location")
ax.set_ylabel("Average Salary ($)")
ax.tick_params(axis="x", rotation=35)
for bar in bars:
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + top_locations.max() * 0.005,
            f"${int(bar.get_height()):,}",
            ha="center", va="bottom", fontsize=8)
plt.tight_layout()
plt.show()
#
```

## 5. Salary vs Numerical Features — Scatter Plots

```python

x_cols = ["experience_years", "skills_count", "certifications"]
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Salary vs Numerical Features", fontsize=14, fontweight="bold")

colors = sns.color_palette("muted", 3)
for i, (col, color) in enumerate(zip(x_cols, colors)):
    sample = df.sample(n=min(5000, len(df)), random_state=42)
    axes[i].scatter(sample[col], sample["salary"],
                    alpha=0.25, s=12, color=color)
    # Trend line
    z = np.polyfit(sample[col], sample["salary"], 1)
    p = np.poly1d(z)
    x_line = np.linspace(sample[col].min(), sample[col].max(), 200)
    axes[i].plot(x_line, p(x_line), color="red", linewidth=1.5, label="Trend")
    axes[i].set_title(f"Salary vs {col.replace('_', ' ').title()}", fontsize=11)
    axes[i].set_xlabel(col.replace("_", " ").title())
    axes[i].set_ylabel("Salary")
    axes[i].legend(fontsize=9)

plt.tight_layout()
plt.show()
```

## 6. Multi-variable Analysis

```python

# Average salary heatmap: Industry x Education Level
pivot = df.pivot_table(values="salary", index="industry",
                       columns="education_level", aggfunc="mean")

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="YlOrRd",
            linewidths=0.4, ax=ax, cbar_kws={"label": "Avg Salary ($)"})
ax.set_title("Average Salary — Industry × Education Level", fontsize=13, fontweight="bold")
ax.set_xlabel("Education Level")
ax.set_ylabel("Industry")
plt.tight_layout()
plt.show()
```

```python

# Salary vs Experience coloured by Education Level + Company Size side-by-side
fig, axes = plt.subplots(1, 2, figsize=(18, 6))
fig.suptitle("Salary vs Experience Years", fontsize=13, fontweight="bold")

sample = df.sample(n=min(4000, len(df)), random_state=0)

# Colored by education level
edu_palette = sns.color_palette("muted", sample["education_level"].nunique())
for idx, (edu, grp) in enumerate(sample.groupby("education_level")):
    axes[0].scatter(grp["experience_years"], grp["salary"],
                    alpha=0.35, s=14, label=edu, color=edu_palette[idx])
axes[0].set_title("Colour = Education Level")
axes[0].set_xlabel("Experience Years")
axes[0].set_ylabel("Salary")
axes[0].legend(title="Education", fontsize=8, title_fontsize=9)

# Colored by company size
size_palette = sns.color_palette("Set2", sample["company_size"].nunique())
for idx, (sz, grp) in enumerate(sample.groupby("company_size")):
    axes[1].scatter(grp["experience_years"], grp["salary"],
                    alpha=0.35, s=14, label=sz, color=size_palette[idx])
axes[1].set_title("Colour = Company Size")
axes[1].set_xlabel("Experience Years")
axes[1].set_ylabel("Salary")
axes[1].legend(title="Company Size", fontsize=8, title_fontsize=9)

plt.tight_layout()
plt.show()
```

```python

# Average salary by job title x remote_work — grouped bar chart
pivot2 = (df.groupby(["job_title", "remote_work"])["salary"]
            .mean()
            .unstack("remote_work"))

fig, ax = plt.subplots(figsize=(16, 6))
pivot2.plot(kind="bar", ax=ax, colormap="muted", edgecolor="white", linewidth=0.5)
ax.set_title("Average Salary by Job Title & Remote Work Policy", fontsize=13, fontweight="bold")
ax.set_xlabel("Job Title")
ax.set_ylabel("Average Salary ($)")
ax.tick_params(axis="x", rotation=35)
ax.legend(title="Remote Work", fontsize=9, title_fontsize=10)
plt.tight_layout()
plt.show()
```
